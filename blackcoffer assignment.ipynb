```python
import os

def create_word_dictionary(positive_file_path, negative_file_path, stop_words_list):
    positive_words = set()
    negative_words = set()

    # Load positive words from file
    with open(positive_file_path, 'r') as positive_file:
        positive_words.update(word.strip() for word in positive_file.readlines())

    # Load negative words from file
    with open(negative_file_path, 'r') as negative_file:
        negative_words.update(word.strip() for word in negative_file.readlines())

    # Exclude stop words
    positive_words -= set(stop_words_list)
    negative_words -= set(stop_words_list)

    return positive_words, negative_words

# Example Usage
positive_file_path = "C:\\Users\\Admin\\Downloads\\Sentimental Analysis\\positive-words.txt"
negative_file_path = "C:\\Users\\Admin\\Downloads\\Sentimental Analysis\\negative-words.txt"
stop_words_list = ["your", "stop", "words", "list"]

positive_words, negative_words = create_word_dictionary(positive_file_path, negative_file_path, stop_words_list)
```

```python
def create_word_dictionary(positive_folder, negative_folder, stop_words_list):
    positive_words = set()
    negative_words = set()
    
    for folder, label_set in [(positive_folder, positive_words), (negative_folder, negative_words)]:
        for filename in os.listdir(folder):
            if filename.endswith(".txt"):
                file_path = os.path.join(folder, filename)
                with open(file_path, 'r') as file:
                    words = [line.strip() for line in file.readlines()]
                    label_set.update(word for word in words if word not in stop_words_list and word.isalpha())
    
    return positive_words, negative_words
```

```python
def calculate_sentiment_scores(tokens, positive_words, negative_words):
    positive_score = sum(1 for word in tokens if word in positive_words)
    negative_score = -1 * sum(1 for word in tokens if word in negative_words)
    polarity_score = (positive_score - negative_score) / (positive_score + negative_score + 0.000001)
    subjectivity_score = (positive_score + negative_score) / (len(tokens) + 0.000001)
    
    return ''positive_score'', negative_score, polarity_score, subjectivity_score
```

```python
# Step 2: Analysis of Readability
def calculate_readability_metrics(text):
    sentences = sent_tokenize(text)
    words = word_tokenize(text)
    average_sentence_length = len(words) / len(sentences)
    
    complex_words = [word for word in words if len(word) > 2]  # Assuming length > 2 implies complexity
    percentage_complex_words = len(complex_words) / len(words)
    
    fog_index = 0.4 * (average_sentence_length + percentage_complex_words)
    
    return average_sentence_length, percentage_complex_words, fog_index
```

```python
# Step 5: Word Count
def count_total_words(cleaned_text):
    stop_words = set(stopwords.words('english'))
    words = [word.strip('?!.,') for word in cleaned_text.split() if word.lower() not in stop_words]
    return len(words)
```

```python
# Step 6: Syllable Count Per Word
def count_syllables_per_word(word):
    vowels = "aeiouy"
    count = 0
    
    if word.endswith("es") or word.endswith("ed"):
        return count
    
    for char in word:
        if char.lower() in vowels:
            count += 1
    
    if count == 0:
        count = 1
    
    return count
```

```python
# Step 7: Personal Pronouns
def count_personal_pronouns(text):
    personal_pronouns = ['I', 'we', 'my', 'ours', 'us']
    pattern = re.compile(r'\b(?:' + '|'.join(personal_pronouns) + r')\b', flags=re.IGNORECASE)
    matches = re.findall(pattern, text)
    count = len(matches)
    
    return count
```
